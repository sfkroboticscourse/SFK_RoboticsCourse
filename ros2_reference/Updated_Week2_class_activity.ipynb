## Core Concept: Nodes and Topics

Before we start the hands-on work, let's understand two fundamental ROS2 concepts.

### What are Nodes?

A **node** is a **single process** (program) in ROS2.

**Examples we'll see today:**
- `joint_state_publisher_gui` - Publishes joint positions when you move sliders
- `robot_state_publisher` - Converts joint positions to 3D transforms
- `rviz` - Visualizes the robot
- **Your custom nodes** - Publisher and subscriber you'll write!

**Key Concept:** Each node does **one specific job**. Complex behavior comes from many nodes working together.

---

### What are Topics?

A **topic** is a **named communication channel** that nodes use to send/receive messages.

**Examples we'll see today:**
- `/joint_states` - Joint positions (angles, velocities)
- `/robot_description` - The URDF content
- `/tf` - Coordinate transforms
- **Your custom topic** - The channel your nodes will use to communicate!

---

### The Publish-Subscribe Pattern

```
+------------------+           +------------------+
|   Publisher      |           |   Subscriber     |
|     Node         |           |      Node        |
+--------+---------+           +--------+---------+
         |                              ^
         |  publishes to                |  subscribes to
         |                              |
         +--------> TOPIC <-------------+
                 /joint_states
```

**Key Points:**
- Nodes **don't talk directly** to each other
- They communicate through **topics**
- **Publishers** send data to topics
- **Subscribers** receive data from topics
- Multiple nodes can subscribe to the same topic
- Multiple nodes can publish to the same topic

### Step 2: Understand the create_publisher and create_subscription Methods

**The create_publisher method takes 3 arguments:**
```python
self.publisher_ = self.create_publisher(
    MessageType,      # 1. What type of data (String, Int32, JointState, etc.)
    'topic_name',     # 2. The channel name - YOU CHOOSE THIS!
    queue_size        # 3. How many messages to buffer (usually 10)
)
```

**The create_subscription method takes 4 arguments:**
```python
self.subscription = self.create_subscription(
    MessageType,         # 1. What type of data (must match publisher!)
    'topic_name',        # 2. The channel name - MUST MATCH PUBLISHER!
    self.callback_func,  # 3. Function to call when message arrives
    queue_size           # 4. How many messages to buffer (usually 10)
)
```

**CRITICAL:** The topic name MUST be **exactly the same** in both nodes!

## Essential Commands Reference

### Node Commands
```bash
ros2 node list                    # List all running nodes
ros2 node info <node_name>        # Get info about a node
```

### Topic Commands
```bash
ros2 topic list                   # List all topics
ros2 topic echo <topic_name>      # View live topic data
ros2 topic info <topic_name>      # Get info about a topic
ros2 topic pub <topic> <type> '<data>'  # Publish to a topic
```

### Running Nodes
```bash
ros2 run <package_name> <node_name>     # Run a node from a package
ros2 launch <package_name> <launch_file>  # Launch a launch file
```

### Interface Commands
```bash
ros2 interface show <msg_type>    # Show message structure
```

### Visualization
```bash
rqt_graph                         # Show node/topic graph
```

### Workspace Commands (Use these frequently!)
```bash
cd ~/sfkrobotics_ws/src                                 # Go to source directory
git pull origin ros2-dev                                # Get latest code
cd ~/sfkrobotics_ws                                     # Go to workspace root
colcon build --packages-select mini_pupper_description  # Build specific package
colcon build --packages-select simple_node              # Build simple_node package
source install/setup.bash                               # Load built packages
```

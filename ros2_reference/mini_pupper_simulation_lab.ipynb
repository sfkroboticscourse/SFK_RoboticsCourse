## Part 1: Understanding the Robot and Its Sensors

### 1.1 The Mini Pupper 2 Robot

The Mini Pupper 2 is a quadruped (four-legged) robot. In our simulation, it is equipped with:

| Sensor | Description | ROS Topic | Use Case |
|--------|-------------|-----------|----------|
| **LiDAR** | 360-degree laser scanner that measures distances to obstacles | `/scan` | Obstacle detection, mapping, navigation |
| **Camera** | Front-facing RGB camera | `/image_raw` | Visual navigation, object recognition |
| **IMU** | Inertial Measurement Unit (accelerometer + gyroscope) | `/imu/data` | Balance, orientation tracking |

### 1.2 What is LiDAR?

**LiDAR** (Light Detection and Ranging) works by:
1. Emitting laser beams in a 360-degree sweep around the robot
2. Measuring how long it takes for each beam to bounce back
3. Calculating the distance to obstacles based on the return time

The Mini Pupper's simulated LiDAR:
- Scans 360 degrees (full circle)
- Has 1040 samples per scan
- Detects objects from 0.2m to 50m away
- Updates at 30 Hz (30 times per second)

### 1.3 What is a Camera?

The camera provides:
- 640x480 pixel RGB images
- 80-degree field of view (horizontal)
- 30 frames per second
- Front-facing perspective (what's ahead of the robot)

## Part 5: Programming the Robot to Exit the Maze

Now for the main challenge! Instead of controlling the robot in real-time with the keyboard, you will **program a sequence of commands** that the robot will execute to navigate through the maze autonomously.

We will use the `ros2 topic pub` command to send movement commands directly from the terminal.

---

### 5.1 Understanding the Command Structure

The robot receives movement commands on the `/cmd_vel` topic. Each command is a `Twist` message with:

- **linear.x**: Forward/backward speed (positive = forward, negative = backward)
- **linear.y**: Left/right strafe (usually 0 for ground robots)
- **linear.z**: Up/down (0 for ground robots)
- **angular.x**: Roll rotation (0 for ground robots)
- **angular.y**: Pitch rotation (0 for ground robots)
- **angular.z**: Yaw rotation (positive = turn left, negative = turn right)

The basic command format is:

```bash
ros2 topic pub --once /cmd_vel geometry_msgs/msg/Twist "{linear: {x: 0.0, y: 0.0, z: 0.0}, angular: {x: 0.0, y: 0.0, z: 0.0}}"
```

---

### 5.2 Basic Movement Commands

Here are the essential commands you'll use:

#### Move Forward
```bash
ros2 topic pub --once /cmd_vel geometry_msgs/msg/Twist "{linear: {x: 0.15, y: 0.0, z: 0.0}, angular: {x: 0.0, y: 0.0, z: 0.0}}"
```

#### Move Backward
```bash
ros2 topic pub --once /cmd_vel geometry_msgs/msg/Twist "{linear: {x: -0.15, y: 0.0, z: 0.0}, angular: {x: 0.0, y: 0.0, z: 0.0}}"
```

#### Turn Left
```bash
ros2 topic pub --once /cmd_vel geometry_msgs/msg/Twist "{linear: {x: 0.0, y: 0.0, z: 0.0}, angular: {x: 0.0, y: 0.0, z: 0.5}}"
```

#### Turn Right
```bash
ros2 topic pub --once /cmd_vel geometry_msgs/msg/Twist "{linear: {x: 0.0, y: 0.0, z: 0.0}, angular: {x: 0.0, y: 0.0, z: -0.5}}"
```

#### Stop (IMPORTANT!)
```bash
ros2 topic pub --once /cmd_vel geometry_msgs/msg/Twist "{linear: {x: 0.0, y: 0.0, z: 0.0}, angular: {x: 0.0, y: 0.0, z: 0.0}}"
```

---

### 5.3 Timed Movement Commands

The `--once` flag sends the command only once. For sustained movement, we need to send the command continuously for a duration. Here's how:

#### Move Forward for ~2 seconds, then stop:
```bash
timeout 2 ros2 topic pub /cmd_vel geometry_msgs/msg/Twist "{linear: {x: 0.15, y: 0.0, z: 0.0}, angular: {x: 0.0, y: 0.0, z: 0.0}}" && ros2 topic pub --once /cmd_vel geometry_msgs/msg/Twist "{linear: {x: 0.0, y: 0.0, z: 0.0}, angular: {x: 0.0, y: 0.0, z: 0.0}}"
```

#### Turn Left for ~1 second, then stop:
```bash
timeout 1 ros2 topic pub /cmd_vel geometry_msgs/msg/Twist "{linear: {x: 0.0, y: 0.0, z: 0.0}, angular: {x: 0.0, y: 0.0, z: 0.5}}" && ros2 topic pub --once /cmd_vel geometry_msgs/msg/Twist "{linear: {x: 0.0, y: 0.0, z: 0.0}, angular: {x: 0.0, y: 0.0, z: 0.0}}"
```

#### Turn Right for ~1 second, then stop:
```bash
timeout 1 ros2 topic pub /cmd_vel geometry_msgs/msg/Twist "{linear: {x: 0.0, y: 0.0, z: 0.0}, angular: {x: 0.0, y: 0.0, z: -0.5}}" && ros2 topic pub --once /cmd_vel geometry_msgs/msg/Twist "{linear: {x: 0.0, y: 0.0, z: 0.0}, angular: {x: 0.0, y: 0.0, z: 0.0}}"
```

### Activity 5.2: Chaining Commands Together

You can chain multiple commands using `&&` to execute them in sequence. For example:

```bash
# Move forward for 2 seconds, then turn left for 1 second, then move forward for 2 seconds
timeout 2 ros2 topic pub /cmd_vel geometry_msgs/msg/Twist "{linear: {x: 0.15, y: 0.0, z: 0.0}, angular: {x: 0.0, y: 0.0, z: 0.0}}" && \
ros2 topic pub --once /cmd_vel geometry_msgs/msg/Twist "{linear: {x: 0.0, y: 0.0, z: 0.0}, angular: {x: 0.0, y: 0.0, z: 0.0}}" && \
sleep 0.5 && \
timeout 1 ros2 topic pub /cmd_vel geometry_msgs/msg/Twist "{linear: {x: 0.0, y: 0.0, z: 0.0}, angular: {x: 0.0, y: 0.0, z: 0.5}}" && \
ros2 topic pub --once /cmd_vel geometry_msgs/msg/Twist "{linear: {x: 0.0, y: 0.0, z: 0.0}, angular: {x: 0.0, y: 0.0, z: 0.0}}" && \
sleep 0.5 && \
timeout 2 ros2 topic pub /cmd_vel geometry_msgs/msg/Twist "{linear: {x: 0.15, y: 0.0, z: 0.0}, angular: {x: 0.0, y: 0.0, z: 0.0}}" && \
ros2 topic pub --once /cmd_vel geometry_msgs/msg/Twist "{linear: {x: 0.0, y: 0.0, z: 0.0}, angular: {x: 0.0, y: 0.0, z: 0.0}}"
```

**Note:** The `\` at the end of each line allows you to split the command across multiple lines for readability. The `sleep 0.5` commands add a small pause between movements for stability.

## Appendix B: Troubleshooting

### Problem: Robot flies or behaves erratically
**Solution:** This is normal! Stop the simulation (Ctrl+C) and restart it. The physics simulation can sometimes become unstable.

### Problem: Commands don't seem to do anything
**Solutions:**
1. Make sure the simulation is running
2. Check that teleop is NOT running (it might be overriding your commands)
3. Verify the topic name is correct: `/cmd_vel`

### Problem: RViz shows no data
**Solutions:**
1. Check that "Fixed Frame" is set to "odom"
2. Verify the simulation is running
3. Check that the topic is correct for LaserScan (`/scan`) and Image (`/image_raw`)

### Problem: Teleop keys don't work
**Solution:** Make sure the terminal running teleop is in focus (click on it). The terminal must be selected for keyboard input to be captured.

### Problem: Robot turns wrong amount
**Solution:** Adjust the timeout duration. A 90-degree turn at angular velocity 0.5 takes approximately 1.6 seconds.

### Problem: Simulation is slow or laggy
**Solution:** This is expected in a web-based environment. Work slowly and give commands time to execute.

## Appendix C: Understanding ROS 2 Topics

ROS 2 uses a "publish-subscribe" communication model:

- **Publishers** send messages to a **topic**
- **Subscribers** receive messages from a **topic**
- Multiple nodes can publish and subscribe to the same topic

### Key Topics for Mini Pupper

| Topic | Type | Description |
|-------|------|-------------|
| `/cmd_vel` | `geometry_msgs/msg/Twist` | Robot velocity commands |
| `/scan` | `sensor_msgs/msg/LaserScan` | LiDAR scan data |
| `/image_raw` | `sensor_msgs/msg/Image` | Camera image data |
| `/imu/data` | `sensor_msgs/msg/Imu` | IMU sensor data |
| `/odom` | `nav_msgs/msg/Odometry` | Robot odometry (position estimate) |
| `/robot_description` | `std_msgs/msg/String` | Robot URDF model |

When you use `ros2 topic pub`, you are acting as a publisher sending messages to the specified topic. When the simulation receives these messages on `/cmd_vel`, it moves the robot accordingly.
